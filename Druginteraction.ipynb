# [dbo].[usp_CheckInteractions]

```sql

SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

CREATE PROCEDURE [dbo].[usp_CheckInteractions]
    @DrugList NVARCHAR(MAX)
AS
BEGIN
    SET NOCOUNT ON;

    ;WITH InputDrugs AS (
        SELECT LTRIM(RTRIM(value)) AS DrugName
        FROM STRING_SPLIT(@DrugList, ',')
        WHERE LTRIM(RTRIM(value)) <> ''
    ),
    MatchedIngredients AS (
        SELECT DISTINCT d.IngredientID
        FROM dbo.Drugs d
        INNER JOIN InputDrugs i
            ON LOWER(d.DrugName) = LOWER(i.DrugName)
        WHERE d.IngredientID IS NOT NULL
    )
    SELECT
        ai1.IngredientName AS IngredientA,
        ai2.IngredientName AS IngredientB,
        i.Severity,
        i.Description
    FROM dbo.Interactions i
    INNER JOIN MatchedIngredients m1
        ON i.Ingredient1ID = m1.IngredientID
    INNER JOIN MatchedIngredients m2
        ON i.Ingredient2ID = m2.IngredientID
    INNER JOIN dbo.Active_Ingredients ai1
        ON ai1.IngredientID = i.Ingredient1ID
    INNER JOIN dbo.Active_Ingredients ai2
        ON ai2.IngredientID = i.Ingredient2ID
    WHERE i.Ingredient1ID < i.Ingredient2ID;
END;
GO
```
